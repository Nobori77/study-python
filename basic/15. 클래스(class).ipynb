```python
# 이름(name), 나이(age), 운동종목(sports)
# 플레이어들을 소개하는 메서드

# 운동선수 3명을 객체화 하고
# 3명을 각각 소개하는 메서드를 사용해서 정보들을 출력하세요.
# ex) OOO선수의 나이 OO살 운동종목 OOOO

class Player:
    def __init__(self, name, age, sports):
        self.name = name
        self.age = age
        self.sports = sports
        
    def introduce(self):
        print(f"{self.name} 선수의 나이는 {self.age}살 운동종목은 {self.sports}")

kim = Player("김철수", 20, "농구")
hong = Player("홍길동", 25, "야구")
park = Player("박지성", 30, "축구")

players = [kim, hong, park]

for player in players:
    player.introduce()
```

```python
# 완전 객체 지향형 언어
# Market
# 판매
class Market:
    def __init__(self, market_name, product_name, product_price):
        self.market_name = market_name
        self.product_name = product_name
        self.product_price = product_price

    def sell(self, customer):
        customer.money -= self.product_price * (100 - customer.discount) / 100

# Customer
# 할인율
class Customer:
    def __init__(self, name, money,discount):
        self.name = name
        self.money = money
        self.discount = discount
```

```python
cu = Market("CU", "도시락", 10000)
customer = Customer("이규혁", 50000, 99)

cu.sell(customer)
print(customer.money)
```

```python
# Atm, User 객체
# deposit

# Atm.deposit(User, 10000)
# 유저의 돈은 10000 차감,
# 은행에 돈은 10000 입금

# 은행이 가진돈과 유저의 돈 출력

class Atm:
    def __init__ (self, atm_money):
        self.atm_money = atm_money

class User:
    def __init__(self, user_money):
        self.user_money = user_money

    def withdraw(self, atm, money):
        self.user_money -= money
        atm.atm_money += money

user = User(1000000)
atm = Atm(20000000)

user.withdraw(atm, 150000)

print(user.user_money)
print(atm.atm_money)
```

```python
class Atm:
    def __init__(self, name):
        self.name = name
        self.users = {}

    def deposit(self, user, money):

        if user.money < money:
            print("잔액 부족")
        
        if user.name in self.users:
            # 기존의 돈
            self.users.update({user.name:  self.users[user.name] + money})
        else:
            self.users.update({user.name: money})

        user.money -= money
        
    def show(self):
        print(self.users)

class User:
    def __init__(self, name, money):
        self.name = name
        self.money = money

    def show(self):
        format_money = format(self.money, "_")
        print(f"{self.name}님의 잔 액 : {format_money}원")
```

```python
yeoksam_atm = Atm("yeoksam")

ggyu = User("이규학", 50000)

yeoksam_atm.deposit(ggyu, 10000)
yeoksam_atm.deposit(ggyu, 10000)
yeoksam_atm.deposit(ggyu, 1000000)
yeoksam_atm.show()
ggyu.show()
```

```python
class Store:
    def __init__(self, name, products, income):
        self.name = name
        self.products = {}
        self.income = 0
        
    # 상품 등록
    def register(self):
        product = input("등록하실 상품명을 입력하세요")
        if product in self.products:
         # 기존의 상품이 있다면 추가하지 않는다.(유효성 검사)
            print("이미 등록된 상품입니다")
            return
        else:
            # 기존의 상품이 없다면 추가
            price = int(input("가격 입력: "))
            stock = int(input("재고 입력: "))

            self.products[product] = {
                "price": price,
                "stock": stock
            }

            print("상품 등록 완료")


    # 상품 판매
    def sell(self, user):
        product = input("구매할 상품명 입력: ")

        if product not in self.products:
            print("없는 상품입니다")
            return

        count = int(input("구매 수량 입력: "))

        price = self.products[product]["price"]
        stock = self.products[product]["stock"]
        total_price = price * count

        if stock < count:
            print("재고가 부족합니다")
            return

        if user.money < total_price:
            print("돈이 부족합니다")
            return

        # 상품의 stock 구매수량만큼 제거
        self.products[product]["stock"] -= count
        # 구매자의 money도 빠져야함
        user.money -= total_price
        # 수익 store 저장
        self.income += total_price
        

        print("상품 판매 완료")
        

    # 상품 수정
    # 이름, 가격, 재고 수정
    def modify(self):
        product = input("수정할 상품명을 입력하세요: ")
    
        # 상품 존재 여부 검사
        if product not in self.products:
            print("존재하지 않는 상품입니다")
            return
    
        new_name = input("새 상품명 입력: ")
        new_price = int(input("새 가격 입력: "))
        new_stock = int(input("새 재고 입력: "))
    
        # 기존 상품 삭제
        del self.products[product]
    
        # 새 정보 저장
        self.products[new_name] = {
            "price": new_price,
            "stock": new_stock
        }
    
        print("상품 수정 완료")
    
    
    # 상품 삭제
    def remove(self):
        product = input("삭제할 상품명을 입력하세요: ")
    
        # 상품 존재 여부 검사
        if product not in self.products:
            print("존재하지 않는 상품입니다")
            return
    
        del self.products[product]
    
        print("상품 삭제 완료")

class User:
    def __init__(self, name, money):
        self.name = name
        self.money = money
```

```python
# 클래스 메서드
class Korean:
    # 클래스 변수
    country = "한국"

    # 클래스 메서드
    # 인스턴스 X, 클래스 변수 O
    @classmethod # 데코레이터
    def trip(cls, country):
        if cls.country == country:
            print("국내 여행")
        else:
            print("해외 여행")

    # 정적 메서드
    @staticmethod
    def slogan():
        print("Imagine your korean")
```

```python
hong = Korean()


hong.trip("한국")
hong.slogan()

Korean.trip("미국")
Korean.slogan()
```

## 상속 (inheritance)

```python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age
```

```python
hong = Person("홍길동")
print(hong)
print(id(hong)) # id 가 실제 메모리에 올라간 주소값

jang = Student("장보고", 20)

# 부모 -> 타입도 같이 전달
# 모든 자식은 부모 타입이다.
# 부모는 자식 타입이 아니다.
print(isinstance(hong, Person))
print(isinstance(hong, Student))
print(isinstance(jang, Person))
print(isinstance(jang, Student))
```

```python
class Animal:
    def __init__(self, name, feed, species):
        self.name = name
        self.feed = feed
        self.species = species

    def feeding(self):
        print(f"{self.name}은 먹이를 먹습니다.")

class Dog(Animal):
    def __init__(self, name, feed, species):
        super().__init__(name, feed, species)

    def act(self):
        print(f"{self.name}은 산책을 합니다.")

class Cat(Animal):
    def __init__(self, name, feed, species):
        super().__init__(name, feed, species)

    def sleep(self):
        print(f"{self.name}은 잠을 자주 잡니다.")
```

```python
dog = Dog("누렁이", "잡식", "강아지")
dog.act()
dog.feeding()

cat = Cat("야옹이", "츄르", "고양이")
cat.sleep()
cat.feeding()
```

```python
# 사람(부모), 학생(자식)
# 사람 - 이름, 나이, 키, 몸무게
# 학생 - 학생, 핸드폰

# 사람은 인사하는 메서드
# 학생은 공부하는 메서드

class Person:
    def __init__(self, name, age, height, weight):
        self.name = name
        self.age = age
        self.height = height
        self.weight = weight

    def greeting(self):
        print(f"{self.name}이 반갑게 인사를 합니다.")

class Student(Person):
    def __init__(self, name, age, height, weight, phone):
        super().__init__(name, age, height, weight)
        self.phone = phone

    def study(self):
        print(f"{self.name}은(는) 열심히 공부를 합니다.")
```

```python
hong = Person("홍길동", 30, "180cm", "70kg")
hong.greeting()
print(hong.name)
print(hong.age)
print(hong.height)
print(hong.weight)

jang = Student("장보고", 18, "170cm", "60kg", "010-1234-1234")
jang.study()
print(jang.name)
print(jang.age)
print(jang.height)
print(jang.weight)
print(jang.phone)
```

```python
class Car:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def start_engine(self):
        print(f"{self.brand}는 열쇠를 시동을 켭니다")

    def stop_engine(self):
        print(f"{self.brand}는 열쇠로 시동을 끕니다")

class SuperCar(Car):
    def __init__(self, brand, price):
        super().__init__(brand, price)

    def start_engine(self):
        print(f"{self.brand}는 음성으로 시동을 켭니다.")

    def stop_engine(self):
        print(f"{self.brand}는 음성으로 시동을 끕니다.")
        
```

```python
matiz = Car("마티즈", 10000000)
matiz.start_engine()
matiz.stop_engine()
```

```python
# 휴대폰 필드
# 브랜드(brand), 가격(price)
# 2G 전화하기
# 3G 전화하기, 인터넷
# 4G 전화하기, 인터넷, 영상통화

class Phone():
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

class Phone_2g(Phone):
    def __init__(self, brand, price, call):
        super().__init__(brand,price)
        self.call = call

class Phone_3g(Phone_2g):
    def __init__(self, brand, price, call, internet):
        super().__init__(brand, price, call)
        self.internet = internet

class Phone_4g(Phone_3g):
    def __init__(self, brand, price, call, internet, video_call):
        super().__init__(brand, price, call, internet)
        self.video_call = video_call
    
```

```python
phone = Phone("삼성", 1000000)
print(phone.brand)
print(phone.price)

phone_2g = Phone_2g("애플", 1500000, True)
print(phone_2g.brand)
print(phone_2g.price)
print(phone_2g.call)

phone_3g = Phone_3g("샤오미", 500000, True, True)
print(phone_3g.brand)
print(phone_3g.price)
print(phone_3g.call)
print(phone_3g.internet)

phone_4g = Phone_4g("모토롤라", 2000000, True, True, True)
print(phone_4g.brand)
print(phone_4g.price)
print(phone_4g.call)
print(phone_4g.internet)
print(phone_4g.video_call)
```

```python
class Phone():
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

class Phone_2g(Phone):
    def __init__(self, brand, price):
        super().__init__(brand,price)

    def call(self):
        print("전화 연결 가능")
        

class Phone_3g(Phone_2g):
    def __init__(self, brand, price):
        super().__init__(brand, price)

    def internet(self):
        print("인터넷 연결 가능")
        

class Phone_4g(Phone_3g):
    def __init__(self, brand, price):
        super().__init__(brand, price)

    def video_call(self):
        print("영상 통화 연결 가능")
```

```python
phone = Phone("삼성", 300000)
print(phone.brand)
print(phone.price)

phone_2g = Phone_2g("애플", 1500000)
print(phone_2g.brand)
print(phone_2g.price)
phone_2g.call()

phone_3g = Phone_3g("샤오미", 500000)
print(phone_3g.brand)
print(phone_3g.price)
phone_3g.call()
phone_3g.internet()

phone_4g = Phone_4g("모토롤라", 700000)
print(phone_4g.brand)
print(phone_4g.price)
phone_4g.call()
phone_4g.internet()
phone_4g.video_call()
```
