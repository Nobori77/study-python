```python
# 이름(name), 나이(age), 운동종목(sports)
# 플레이어들을 소개하는 메서드

# 운동선수 3명을 객체화 하고
# 3명을 각각 소개하는 메서드를 사용해서 정보들을 출력하세요.
# ex) OOO선수의 나이 OO살 운동종목 OOOO

class Player:
    def __init__(self, name, age, sports):
        self.name = name
        self.age = age
        self.sports = sports
        
    def introduce(self):
        print(f"{self.name} 선수의 나이는 {self.age}살 운동종목은 {self.sports}")

kim = Player("김철수", 20, "농구")
hong = Player("홍길동", 25, "야구")
park = Player("박지성", 30, "축구")

players = [kim, hong, park]

for player in players:
    player.introduce()
```

```python
# 완전 객체 지향형 언어
# Market
# 판매
class Market:
    def __init__(self, market_name, product_name, product_price):
        self.market_name = market_name
        self.product_name = product_name
        self.product_price = product_price

    def sell(self, customer):
        customer.money -= self.product_price * (100 - customer.discount) / 100

# Customer
# 할인율
class Customer:
    def __init__(self, name, money,discount):
        self.name = name
        self.money = money
        self.discount = discount
```

```python
cu = Market("CU", "도시락", 10000)
customer = Customer("이규혁", 50000, 99)

cu.sell(customer)
print(customer.money)
```

```python
# Atm, User 객체
# deposit

# Atm.deposit(User, 10000)
# 유저의 돈은 10000 차감,
# 은행에 돈은 10000 입금

# 은행이 가진돈과 유저의 돈 출력

class Atm:
    def __init__ (self, atm_money):
        self.atm_money = atm_money

class User:
    def __init__(self, user_money):
        self.user_money = user_money

    def withdraw(self, atm, money):
        self.user_money -= money
        atm.atm_money += money

user = User(1000000)
atm = Atm(20000000)

user.withdraw(atm, 150000)

print(user.user_money)
print(atm.atm_money)
```

```python
class Atm:
    def __init__(self, name):
        self.name = name
        self.users = {}

    def deposit(self, user, money):

        if user.money < money:
            print("잔액 부족")
        
        if user.name in self.users:
            # 기존의 돈
            self.users.update({user.name:  self.users[user.name] + money})
        else:
            self.users.update({user.name: money})

        user.money -= money
        
    def show(self):
        print(self.users)

class User:
    def __init__(self, name, money):
        self.name = name
        self.money = money

    def show(self):
        format_money = format(self.money, "_")
        print(f"{self.name}님의 잔 액 : {format_money}원")
```

```python
class Store:
    def __init__(self, name, products, income):
        self.name = name
        self.products = {}
        self.income = 0
        
    # 상품 등록
    def register(self):
        product = input("등록하실 상품명을 입력하세요")
        if product in self.products:
         # 기존의 상품이 있다면 추가하지 않는다.(유효성 검사)
            print("이미 등록된 상품입니다")
            return
        else:
            # 기존의 상품이 없다면 추가
            price = int(input("가격 입력: "))
            stock = int(input("재고 입력: "))

            self.products[product] = {
                "price": price,
                "stock": stock
            }

            print("상품 등록 완료")


    # 상품 판매
    def sell(self, user):
        product = input("구매할 상품명 입력: ")

        if product not in self.products:
            print("없는 상품입니다")
            return

        count = int(input("구매 수량 입력: "))

        price = self.products[product]["price"]
        stock = self.products[product]["stock"]
        total_price = price * count

        if stock < count:
            print("재고가 부족합니다")
            return

        if user.money < total_price:
            print("돈이 부족합니다")
            return

        # 상품의 stock 구매수량만큼 제거
        self.products[product]["stock"] -= count
        # 구매자의 money도 빠져야함
        user.money -= total_price
        # 수익 store 저장
        self.income += total_price
        

        print("상품 판매 완료")
        

    # 상품 수정
    # 이름, 가격, 재고 수정
    def modify(self):
        product = input("수정할 상품명을 입력하세요: ")
    
        # 상품 존재 여부 검사
        if product not in self.products:
            print("존재하지 않는 상품입니다")
            return
    
        new_name = input("새 상품명 입력: ")
        new_price = int(input("새 가격 입력: "))
        new_stock = int(input("새 재고 입력: "))
    
        # 기존 상품 삭제
        del self.products[product]
    
        # 새 정보 저장
        self.products[new_name] = {
            "price": new_price,
            "stock": new_stock
        }
    
        print("상품 수정 완료")
    
    
    # 상품 삭제
    def remove(self):
        product = input("삭제할 상품명을 입력하세요: ")
    
        # 상품 존재 여부 검사
        if product not in self.products:
            print("존재하지 않는 상품입니다")
            return
    
        del self.products[product]
    
        print("상품 삭제 완료")

class User:
    def __init__(self, name, money):
        self.name = name
        self.money = money
```
