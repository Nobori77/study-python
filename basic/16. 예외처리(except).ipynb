## 커스텀 예외(Custom Exception)

```python
# Exception: 모든 예외의 부모 클래스
class BadWordException(Exception):
    def __init__(self, message):
        super().__init__(message)
```

```python
# 사용자가 "바보" 라고 입력을 하면 커스텀 예외 발생
try:
    expression = input("채팅을 입력하세요")
    
    if "바보" in expression:
        raise BadWordException("바보라고 하지마세요")
        
except BadWordException as e:
    print(e)
```

```python
# 국어, 영어, 수학 점수를 받아서 총점과 평균을 구해주는 프로그램
# 0 ~ 100점 까지만 입력 가능
# 이름, 학번 선택 입력 가능
# 평균 소수점 자리수 선택 가능
# 예외 처리 사용

class BadScoreException(Exception):
    def __init__(self, message):
        super().__init__(message)


class Exam:
    def __init__(self, kor, eng, math, name=None, student_id=None):
        self.kor = kor
        self.eng = eng
        self.math = math
        self.name = name
        self.student_id = student_id

        # 점수 유효성 검사
        for score in [kor, eng, math]:
            if score < 0 or score > 100:
                raise BadScoreException("점수는 0 ~ 100까지만 입력 가능합니다.")

    # 총점
    def get_total(self):
        return self.kor + self.eng + self.math

    # 평균
    def get_average(self, digit=2):
        return round(self.get_total() / 3, digit)

    # 출력
    def show_info(self, digit=2):
        total = self.get_total()
        average = self.get_average(digit)

        if self.name and self.student_id:
            print(f"학번 : {self.student_id}")
            print(f"이름 : {self.name}")

        elif self.name:
            print(f"이름 : {self.name}")

        print(f"국어 : {self.kor}")
        print(f"영어 : {self.eng}")
        print(f"수학 : {self.math}")
        print(f"총점 : {total}")
        print(f"평균 : {average}")


try:
    name = input("이름 입력(없으면 엔터) : ")
    student_id = input("학번 입력(없으면 엔터) : ")

    kor = int(input("국어 점수 : "))
    eng = int(input("영어 점수 : "))
    math = int(input("수학 점수 : "))

    digit = int(input("평균 소수점 자리수 : "))

    exam = Exam(
        kor,
        eng,
        math,
        name if name != "" else None,
        student_id if student_id != "" else None
    )

    exam.show_info(digit)

except ValueError:
    print("숫자를 올바르게 입력하세요.")

except BadScoreException as e:
    print(e)

except Exception as e:
    print("오류 발생 :", e)
```

```python
# 실습
# 국어, 영어, 수학 점수를 받아서 총점과 평균을 구해주는 프로그램
# 0 ~ 100점 까지만 입력할 수 있습니다.
# 이름을 입력할 수도 있습니다.
# 학번을 입력할 수도 있습니다.
# 평균의 소숫점 자리를 입력받아 처리할 수도 있습니다.
# 예외가 발생한다면 try ~ except 문법으로 해결하세요.

class Student:
    def __init__(self, student_id, name, kor, eng, math):
        self.student_id = student_id
        self.name = name
        self.kor = kor
        self.eng = eng
        self.math = math
        self.total = kor + eng + math
        self.average = self.total / 3
    
    def print_score(self, **kwards):
        is_name_print = kwards.get("name", False)
        round_number = kwards.get("round", 0)

        name = f"{self.name}님의 " if is_name_print else ""
        average = round(self.average, round_number) if round_number else int(self.average)
        message = f"{name}총 점수는: {self.total}점, 평균은: {average}점"
        print(message)
```

```python
hong = Student(1, "홍길동", 40, 50, 67)

hong.print_score(name=True, round=3)
```

```python
class ScoreOutOfRangeException(Exception):
    def __init__(self, message):
        super().__init__(message)
```

```python
try:
    kor = int(input("국어 점수를 입력하세요.\nex) 75"))
    math = int(input("수학 점수를 입력하세요.\nex) 75"))
    eng = int(input("영어 점수를 입력하세요.\nex) 75"))
    
    if not (0 <= kor <= 100) or not (0 <= math <= 100) or not (0 <= eng <= 100):
        raise ScoreOutOfRangeException("0 ~ 100까지의 점수를 입력하세요.")

    hong = Student(1, "홍길동", kor, math, eng)

    is_name = eval(input("이름을 출력할까요?\nex)True\n"))
    is_round = int(input("소수점 자리 출력값을 입력하세요.\nex)0"))
    payload = {}
    
    if is_name:
        payload.update({"name": is_name})

    payload.update({"round": is_round})

    hong.print_score(**payload)

except ScoreOutOfRangeException as e:
    print(e)

except ValueError:
    print("타입이 잘못 되었습니다.")
```
