## 1. 리스트 컬렉션 메서드

```python
li1 = [i + 1 for i in range(10)] 
```

```python
# 맨뒤에 추가
li1.append(10)
```

```python
li1
```

```python
for i in range(11, 21):
    li1.append(i)

li1 = [i for i in range(1,21)]
```

```python
li1
```

```python
# list의 모든 합을 구하기
# max(): 가장 큰 값을 찾아서 반환하는 함수
# min(): 가장 작은 값을 찾아서 반환하는 함수
# sum(): 모든 값을 더하는 함수

sum(li1)
max(li1)
min(li1)
```
