## 값을 몇개나 전달할 지 모를 때

```python
# 값을 몇 개를 전달해도 모두 더하는 함수
# args: 가변인자, 가변매개변수, rest parameter

def total(*args):
    print(args)
    print(type(args))
    return sum(args)
```

```python
total(10, 20)
```

```python
# 두 수를 전달하면 사칙연산의 결과 반환
def calc(num1, num2):
    return num1 + num2, num1 - num2, num1 * num2, num1 / num2
```

```python
add, sub, mul, div  = calc(10, 20)
```

```python
print(add)
print(sub)
print(mul)
print(div)
```

## keywards argument

```python
def order_info(*args, name, address, phone):
    total = sum(args)
    print(f"{name}님의 주소: {address}, 연락처: {phone}")
    print(f"총 구매 가격 : {total}")
```

```python
# 이름, 주소, 연락처, 총 구매가격
order_info(1000, 5000, 100000, 500, name="홍길동", address="강남", phone="01012341234")
```

```python
user_info = {
    "name": "홍길동",
    "address": "강남",
    "phone": "01012341234"
}
```

```python
user_info
```

```python
order_info(1000, 5000, 100000, 500, **user_info)
```

```python
def order_info(*args, **keywards):
    # key:value -> dict
    if "name" in keywards:
        print(keywards["name"])
```

```python
order_info(a="10", b="20", c="30")
```

```python
order_info(name="홍길동")
```

```python
# 국어, 영어, 수학 점수 평균 구하기
# 자릿수는 2번째 자리부터 반올림
# 사용자의 정보가 입력되면 같이 출력하기
def get_average(kor, eng, math, name=None):
    total = sum([kor, eng, math])
    average = round(total / 3, 2)
    if name != None:
        print(f"{name}님의 점수는\n국어 : {kor}\n영어: {eng}\n수학: {math}\n총 점수: {total}\n평균: {average}")
    else:
        print(f"평균점수는 {average} 입니다")
```

```python
def get_average1(**kwards):
    kor, eng, math = kwards.get("kor", 0), kwards.get("eng", 0), kwards.get("math", 0)
    total = kor + eng + math

    average = total / 3

    if "round" in kwards:
        average = round(average, kwards["round"])
    
    if "name" in kwards:
        name = kwards["name"]
        return f"{name}님의 평균은: {average}"

    return f"평균은: {average}"
```

```python
get_average(60, 30, 40, "노민균")
```

```python
user_info = {
    "kor": 100,
    "eng": 40,
    "math": 22,
    "name": "홍길동",
    "round": 5
}
```

```python
get_average1(**user_info)
```
